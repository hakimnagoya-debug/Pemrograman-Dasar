```python
# Soal 1 — Kalkulator faktorial 1 sampai 5
n = int(input("Masukkan bilangan bulat (1–5): "))

# Validasi input
if n < 1 or n > 5:
    print("Input tidak valid. Masukkan bilangan antara 1 sampai 5.")

# Perhitungan faktorial
elif n == 1:
    print("1! = 1")

elif n == 2:
    print("2! =", 2 * 1)

elif n == 3:
    print("3! =", 3 * 2 * 1)

elif n == 4:
    print("4! =", 4 * 3 * 2 * 1)

else:
    print("5! =", 5 * 4 * 3 * 2 * 1)
```
