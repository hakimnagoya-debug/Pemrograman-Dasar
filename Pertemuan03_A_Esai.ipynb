```python
jawaban_2 = """
SOAL 2
Operator relasional digunakan untuk membandingkan dua nilai.
Hasil dari perbandingan tersebut berupa True atau False.

Contoh operator relasional:
    >     lebih besar
    <     lebih kecil
    >=    lebih besar atau sama dengan
    <=    lebih kecil atau sama dengan
    ==    sama dengan
    !=    tidak sama dengan

Sedangkan operator logika digunakan untuk menggabungkan atau
mengolah beberapa kondisi. Operator logika yang umum digunakan
adalah:

    and
    or
    not

Contoh situasi nyata adalah menentukan apakah seseorang boleh
mengemudi sendiri. Syaratnya adalah umur minimal 17 tahun dan
memiliki SIM.

Contoh ekspresi:

    (umur >= 17) and punya_sim

Pada ekspresi tersebut, umur >= 17 merupakan operator relasional,
sedangkan and merupakan operator logika.

Contoh:
    umur = 20
    punya_sim = True

Maka:
    (20 >= 17) and True
    True and True
    True

Jadi operator relasional digunakan untuk membuat kondisi,
sedangkan operator logika digunakan untuk menggabungkan kondisi.

SUMBER:
Python Documentation - Comparisons dan Boolean Operations.
"""

print("\n" + "=" * 60)
print(jawaban_2)
```

```python
umur = 20
punya_sim = True

# Operator relasional
syarat_umur = umur >= 17

# Operator relasional + logika
boleh_mengemudi = (umur >= 17) and punya_sim

print("Umur =", umur)
print("Memenuhi umur minimal 17 tahun:", syarat_umur)
print("Memiliki SIM:", punya_sim)
print("Boleh mengemudi sendiri:", boleh_mengemudi)
```

```python
jawaban_3 = """
SOAL 3
Salah satu contoh kesalahan operator precedence dapat ditemukan
pada kasus di Stack Overflow yang menggunakan assignment expression
(:=) dalam kondisi while.

Kode yang bermasalah:

    while food2 := input("Which food do you like? ") != "quit":
        foods2.append(food2)

Program tersebut tidak menyimpan teks makanan ke dalam variabel
food2. Sebaliknya, variabel food2 mendapatkan nilai True atau
False karena operator != dikerjakan terlebih dahulu berdasarkan
aturan precedence.

Ekspresi tersebut secara efektif dibaca sebagai:

    food2 := (input("Which food do you like? ") != "quit")

Akibatnya, ketika pengguna memasukkan makanan yang bukan "quit",
hasil perbandingan menjadi True dan nilai True yang disimpan.

Perbaikannya adalah menggunakan tanda kurung:

    while (food2 := input("Which food do you like? ")) != "quit":
        foods2.append(food2)

Dengan tanda kurung, input disimpan terlebih dahulu ke food2,
kemudian food2 dibandingkan dengan "quit".

Kesalahan ini menunjukkan bahwa penggunaan tanda kurung penting
untuk memastikan urutan operasi sesuai dengan yang diinginkan.

SUMBER:
Stack Overflow - Why does using an assignment expression in a
loop condition store a bool value?
"""

print("\n" + "=" * 60)
print(jawaban_3)



# CONTOH CODING OPERATOR PRECEDENCE


print("\n" + "=" * 60)
print("CONTOH OPERATOR PRECEDENCE")
print("=" * 60)

hasil_1 = 10 + 5 * 2
hasil_2 = (10 + 5) * 2

print("10 + 5 * 2   =", hasil_1)
print("(10 + 5) * 2 =", hasil_2)
```
