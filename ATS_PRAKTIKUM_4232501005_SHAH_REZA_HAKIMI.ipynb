```python
# SOAL NO 1
#a.Energi Harian

daya_kw = 450
jam = 8

energi = daya_kw * jam

print(f"Energi harian = {energi} kWh")


#b.Capacity Factor

energi_aktual = 7200
kapasitas_kw = 400

capacity_factor = (energi_aktual / (kapasitas_kw * 24)) * 100

print(f"Capacity factor = {capacity_factor:.2f}%")
```

Program ini digunakan untuk menghitung energi harian dan capacity factor pada PLTD. Diawal, kita memasukkan nilai daya, lama operasi, energi aktual, dan kapasitas pembangkit ke dalam variabel. Kemudian energi harian dihitung dengan mengalikan daya 450 kW dengan waktu operasi 8 jam, sehingga diperoleh hasil 3.600 kWh.

Selanjutnya capacity factor dihitung dengan membagi energi aktual 7.200 kWh dengan kapasitas 400 kW dikalikan 24 jam, dikalikan juga 100%. Hasilnya adalah 75%. Fungsi print() digunakan untuk menampilkan hasil, sedangkan f-string memudahkan kita memasukkan nilai variabel ke dalam teks. Format :.2f digunakan agar angka ditampilkan di dua angka di belakang koma.

Jadi, hasil akhirnya adalah energi harian sebesar 3.600 kWh dan capacity factor sebesar 75,00%.

```python
# SOAL NO 2
# Data tekanan uap (bar)
tekanan = [12.4, 13.1, 12.8, 14.2, 13.5]

#a.tekanan pada pembacaan ke-2
print(f"Tekanan pembacaan ke-2 = {tekanan[1]} bar")

#b.dua pembacaan terakhir menggunakan slicing
print(f"Dua pembacaan terakhir = {tekanan[-2:]}")

#c.dua pembacaan baru
tekanan.append(13.9)
tekanan.append(12.6)

print(f"Jumlah data setelah penambahan = {len(tekanan)}")
```

Program ini digunakan untuk mengolah data tekanan uap turbin. Pertama, data tekanan dimasukkan ke dalam list bernama tekanan. Untuk menampilkan pembacaan kedua digunakan tekanan[1], Dikarenakan indeks Python dimulai dari 0. Kemudian, tekanan[-2:] digunakan untuk mengambil dua data terakhir.

Selanjutnya fungsi append() digunakan untuk menambahkan dua data baru, yaitu 13.9 dan 12.6. Terakhir, fungsi len() digunakan untuk menghitung jumlah seluruh data setelah ditambahkan.

Jadi hasilnya adalah tekanan pembacaan kedua 13.1 bar, dua pembacaan terakhir 14.2 dan 13.5 bar, dan jumlah data setelah penambahan menjadi 7 data.

```python
# SOAL N0 3
# suhu dan tekanan boiler
suhu = float(input("Masukkan suhu uap (°C): "))
tekanan = float(input("Masukkan tekanan uap (bar): "))

# status operasi boiler
if suhu >= 550 or tekanan >= 14:
    print("STATUS: TRIP")
elif suhu > 520 and tekanan > 12:
    print("STATUS: WARNING")
else:
    print("STATUS: NORMAL")
```

Program ini digunakan untuk menentukan kondisi boiler berdasarkan suhu dan tekanan uap. Pertama, kita memasukkan nilai suhu dan tekanan menggunakan input() dan float() agar data terbaca sebagai angka.

Selanjutnya, program menggunakan if-elif-else untuk menentukan status boiler. Jika suhu mencapai 550°C atau tekanan mencapai 14 bar, statusnya TRIP. Jika suhu lebih dari 520°C dan tekanan lebih dari 12 bar, statusnya WARNING. Selain itu, statusnya NORMAL.

Operator or berarti salah satu kondisi terpenuhi, sedangkan and berarti kedua kondisi harus terpenuhi. Pemeriksaan TRIP dilakukan terlebih dahulu karena merupakan kondisi paling berbahaya.




```python
# SOAL NO 4
# Volume awal tangki
volume = 100

while volume > 0:
    volume -= 20

    print(f"Volume: {volume} liter")

    if volume == 20:
        print("WARNING: Hampir kosong!")
# Pesan tangki kosong
print("TANGKI KOSONG - Pengosongan berhenti.")
```

Program ini digunakan untuk menyimulasikan pengosongan tangki bahan bakar dari 100 liter sampai habis. Pertama, volume diatur sebesar 100 liter. Kemudian, perulangan while digunakan untuk mengurangi bahan bakar sebanyak 20 liter setiap putaran dan menampilkan sisa volumenya.

Jika volume tersisa 20 liter, program menampilkan peringatan "WARNING: Hampir kosong!". Setelah volume mencapai 0 liter, perulangan berhenti dan muncul pesan "TANGKI KOSONG". Jadi, program ini membantu menggambarkan proses pengosongan tangki secara bertahap.

```python
# SOAL NO 5
# Data suhu bearing
data = ["62.5", "58.0", "81.3", "N/A", "66.0"]

jumlah_valid = 0

# Memeriksa setiap data dengan for
for i, nilai in enumerate(data, start=1):

    # Mengubah data menjadi float
    try:
        suhu = float(nilai)
        jumlah_valid += 1

        #pemeriksaan suhu bearing
        if suhu > 70.0:
            print(f"Warning: Titik {i} suhu bearing tinggi!")

    #kalau data tidak valid
    except ValueError:
        print(f"Titik {i}: data tidak valid - pemeriksaan dihentikan")
        break
#jumlah data valid
print(f"Jumlah data valid: {jumlah_valid}")
```

Program ini digunakan untuk memeriksa suhu bearing pada lima titik pengukuran. Data diperiksa satu satu menggunakan perulangan for, lalu diubah menjadi angka dengan float(). Jika suhu lebih dari 70°C, program akan menampilkan peringatan.

Jika terdapat data yang tidak valid seperti "N/A", try-except akan menangani kesalahan dan break menghentikan pemeriksaan. Variabel jumlah_valid digunakan untuk menghitung data yang berhasil diperiksa.

Jadi terdapat 3 data valid dan pemeriksaan berhenti di titik ke-4 karena sensor mengalami kerusakan.

```python
# Program Monitoring Konsumsi Listrik Harian
konsumsi = []

try:
    jumlah_hari = int(input("Masukkan jumlah hari: "))

    if jumlah_hari <= 0:
        print("Jumlah hari harus lebih dari 0!")

    else:
#data konsumsi listrik
        for i in range(jumlah_hari):
            energi = float(input(f"Konsumsi hari ke-{i+1} (kWh): "))

            if energi < 0:
                print("Data tidak boleh negatif!")
                break
            konsumsi.append(energi)
 #total dan rata-rata
        if len(konsumsi) == jumlah_hari:
            total = sum(konsumsi)
            rata = total / jumlah_hari

            print("\nTotal konsumsi:", total, "kWh")
            print("Rata-rata:", rata, "kWh")
#kategori penggunaan listrik
            if rata < 5:
                print("Kategori: HEMAT")
            elif rata <= 10:
                print("Kategori: NORMAL")
            else:
                print("Kategori: BOROS")

except ValueError:
    print("Error! Masukkan angka yang benar.")
```
