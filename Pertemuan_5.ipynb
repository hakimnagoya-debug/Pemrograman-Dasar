```python
# Data belanja buku
harga_buku = 18500
jumlah_buku = 7
diskon_persen = 10
uang_budi = 150000
jumlah_orang = 4

# Menghitung total belanja dan pembayaran
total = jumlah_buku * harga_buku
diskon = total * diskon_persen // 100
total_bayar = total - diskon

# Menghitung kembalian dan pembagian biaya
kembalian = uang_budi - total_bayar
biaya_orang = total_bayar // jumlah_orang
sisa_biaya = total_bayar % jumlah_orang

# Memeriksa kecukupan uang
uang_cukup = uang_budi >= total_bayar

# Menampilkan hasil
print("Total       :", total)
print("Diskon      :", diskon)
print("Bayar       :", total_bayar)
print("Kembalian   :", kembalian)
print("Per orang   :", biaya_orang)
print("Sisa        :", sisa_biaya)
print("Uang cukup? :", uang_cukup)
```

```python
# Data Input
DATA_PLTS = {
    "jumlah_panel": 12,
    "daya_panel": 450,       # Wp
    "jam_sinar": 5,          # jam/hari
    "efisiensi": 0.8,
    "kebutuhan": 25,         # kWh/hari
    "tarif": 1444.70,        # Rp/kwh
    "hari_per_bulan": 30
}

def hitung_plts(data):
    # 1. Energi harian (kWh)
    energi_harian = data["jumlah_panel"] * data["daya_panel"] * data["jam_sinar"] * data["efisiensi"] / 1000

    # 2. Energi bulanan dan nilai rupiah
    energi_bulanan = energi_harian * data["hari_per_bulan"]
    nilai_listrik = energi_bulanan * data["tarif"]

    # 3. Kekurangan energi per hari
    kekurangan = data["kebutuhan"] - energi_harian

    # 4. Apakah mencukupi?
    mencukupi = energi_harian >= data["kebutuhan"]

    return energi_harian, energi_bulanan, nilai_listrik, kekurangan, mencukupi

# Eksekusi perhitungan
e_harian, e_bulanan, n_listrik, krg, cukup = hitung_plts(DATA_PLTS)

# Menampilkan Output
print("Energi harian (kWh):", round(e_harian, 2))
print("Energi bulanan (kWh):", round(e_bulanan, 2))
print("Nilai listrik (Rp) :", round(n_listrik, 2))
print("Kekurangan (kWh)   :", round(krg, 2))
print("Mencukupi?         :", cukup)

```

```python
# Data
daya_generator = 200     # kW
jam_operasi = 8          # jam/hari
konsumsi_solar = 0.25    # liter/kWh
harga_solar = 13500      # Rp/Liter
kapasitas_tangki = 1000  # Liter
arus = 20                # A
hambatan = 0.5           # Ohm

def hitung_generator():
    energi = daya_generator * jam_operasi
    solar = energi * konsumsi_solar
    biaya = solar * harga_solar

    tangki_cukup = kapasitas_tangki // solar
    sisa_solar = kapasitas_tangki % solar

    rugi_kabel = (arus ** 2) * hambatan
    rugi_tinggi = rugi_kabel > 150 and arus >= 20

    return energi, solar, biaya, tangki_cukup, sisa_solar, rugi_kabel, rugi_tinggi

energi, solar, biaya, tangki_cukup, sisa_solar, rugi_kabel, rugi_tinggi = hitung_generator()

print("Energi (kWh)    :", energi)
print("Solar (liter)   :", float(solar))
print("Biaya (Rp)      :", float(biaya))
print("Tangki cukup    :", float(tangki_cukup), "hari")
print("Sisa solar (L)  :", float(sisa_solar))
print("Rugi kabel (W)  :", float(rugi_kabel))
print("Rugi tinggi?    :", rugi_tinggi)
```
