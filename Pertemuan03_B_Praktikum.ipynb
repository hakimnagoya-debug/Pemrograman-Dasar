```python

# RPE311 DASAR PEMROGRAMAN
# PERTEMUAN 3 - EXPRESSIONS & OPERATORS
# Nama : Shah Reza Hakimi
# NIM  : Isi NIM
# Tanggal    : 23 Sepetember 2026

# Input dua bilangan
bilangan1 = 20
bilangan2 = 6

# Operasi aritmatika
penjumlahan = bilangan1 + bilangan2
pengurangan = bilangan1 - bilangan2
perkalian = bilangan1 * bilangan2
pembagian = bilangan1 / bilangan2
pembagian_bulat = bilangan1 // bilangan2
modulus = bilangan1 % bilangan2

# Menampilkan hasil
print("===== HASIL OPERASI ARITMATIKA =====")
print(f"Bilangan pertama = {bilangan1}")
print(f"Bilangan kedua   = {bilangan2}")
print(f"Penjumlahan      = {penjumlahan}")
print(f"Pengurangan      = {pengurangan}")
print(f"Perkalian        = {perkalian}")
print(f"Pembagian        = {pembagian:.2f}")
print(f"Pembagian bulat  = {pembagian_bulat}")
print(f"Modulus          = {modulus}")
```

```python

# Nama       : Shah Reza Hakimi
# NIM        : 4232501005
# Tanggal    : 23 Sepetember 2026
# Tujuan     : Menggunakan operator logika

umur = int(input("Masukkan umur: "))

sim_input = input("Apakah memiliki SIM? (True/False): ")
punya_sim = sim_input == "True"

# Operator relasional + logika
boleh_mengemudi = (umur >= 17) and punya_sim

print("\n=== HASIL ===")
print(f"Umur memenuhi syarat: {umur >= 17}")
print(f"Memiliki SIM: {punya_sim}")
print(f"Boleh mengemudi sendiri: {boleh_mengemudi}")
```

```python

# Nama       : Shah Reza Hakimi
# NIM        : 4232501005
# Tanggal    : 23 Sepetember 2026
# Tujuan     : Menghitung efisiensi termal

energi_input = float(input("Masukkan energi input (MJ): "))
energi_output = float(input("Masukkan energi output (MJ): "))

# Menghitung efisiensi
efisiensi = (energi_output / energi_input) * 100

# Menampilkan hasil dengan 2 angka desimal
print(f"Efisiensi termal unit pembangkit = {efisiensi:.2f}%")
```

```python

# Nama       : Shah Reza Hakimi
# NIM        : 4232501005
# Tanggal    : 23 Sepetember 2026
# Tujuan     : Mengecek kondisi maintenance

suhu_boiler = float(input("Masukkan suhu boiler (°C): "))
beban_operasi = float(input("Masukkan beban operasi (%): "))

# Operator relasional dan logika
perlu_maintenance = (suhu_boiler > 550) or (beban_operasi > 95)

# Menampilkan hasil
print(
    f"Suhu boiler = {suhu_boiler:.2f}°C, "
    f"beban operasi = {beban_operasi:.2f}%, "
    f"perlu maintenance: {perlu_maintenance}"
)
```
