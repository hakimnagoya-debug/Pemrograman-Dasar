```python
tegangan_harian = [217.5, 219.0, 220.3, 218.7, 221.1, 216.9]

# (a) Cetak data pertama dan terakhir
print("Pembacaan pertama:", tegangan_harian[0])
print("Pembacaan terakhir:", tegangan_harian[-1])

# (b) Tambahkan pembacaan baru
tegangan_harian.append(222.0)

# (c) Cetak jumlah total data
print("Jumlah data:", len(tegangan_harian))

# Cetak seluruh data
print("Data tegangan:", tegangan_harian)
```

```python
pembangkit =[
    {
        "nama": "Unit-1",
        "kapasitas_MW": 100,
        "status": "operasi"
    },
    {
        "nama": "Unit-2",
        "kapasitas_MW": 150,
        "status": "operasi"
    },
    {
        "nama": "Unit-3",
        "kapasitas_MW": 80,
        "satus": "operasi"
    }
]

# Cetak nama unit kedua
print("Nama unit kedua:", pembangkit[1]["nama"])

# Ubah status unit-3
pembangkit[2]["status"] = "operasi"

# Tambahkan unit baru
pembangkit.append({
    "nama": "Unit-4",
    "kapasitas_MW": 120,
    "status": "operasi"
})

# Cetak seluruh data
print("\nData seluruh pembangkit:")
print(pembangkit)
```
